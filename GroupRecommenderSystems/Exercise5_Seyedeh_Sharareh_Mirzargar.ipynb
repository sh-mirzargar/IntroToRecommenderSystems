```python
def collect_rating(users_id,user_movie_ratings,rated_movies):
  #for each movie in the list of the rated_movies:
  users_rating=[]

  for movie_idx in range(len(rated_movies)):
    users_rating.append([])
  #iterate over the memberd of the group
    for id in users_id:
      user = "U"+str(id)
      users_rating[movie_idx].append(user_movie_ratings.iloc[movie_idx][user])
  return users_rating
```

```python
def average_strategy(users_rating,num_rated_movies):
  average_ratings = []
  for movie in range(num_rated_movies):
    average_ratings.append(np.mean(np.array(users_rating[movie])))
  return np.array(average_ratings)
```

```python
def multiplicative_strategy(users_rating,num_rated_movies):
  multiplicative_ratings = []
  for movie in range(num_rated_movies):
    multiplicative_ratings.append(np.prod(users_rating[movie]))
  return np.array(multiplicative_ratings)
```

```python
def most_pleasure_strategy(users_rating,num_rated_movies):
  most_pleasure_ratings = []
  for movie in range(num_rated_movies):
    most_pleasure_ratings.append(np.max(np.array(users_rating[movie])))
  return np.array(most_pleasure_ratings)
```

```python
def least_misery_strategy(users_rating,num_rated_movies):
  least_misery_ratings = []
  for movie in range(num_rated_movies):
    least_misery_ratings.append(np.min(np.array(users_rating[movie])))
  return np.array(least_misery_ratings)
```

```python
def average_without_misery_strategy(users_rating,num_rated_movies):
  average_without_misery_ratings = []
  for movie in range(num_rated_movies):
    tmp = np.array(users_rating[movie])
    min = np.min(tmp)
    sum = 0
    for t in tmp:
      if(t!=min):
        sum += t
    average_without_misery_ratings.append(sum/(len(tmp)-1))
  return np.array(average_without_misery_ratings)
```

```python
for key,value in groups.items():
  ratings = {}
  users_rating = collect_rating(value,user_movie_ratings,rated_movies)
  ratings['average'] = average_strategy(users_rating,num_rated_movies)
  ratings['multiplicative'] = multiplicative_strategy(users_rating,num_rated_movies)
  ratings ['most pleasure'] = most_pleasure_strategy(users_rating,num_rated_movies)
  ratings ['least misery strategy'] = least_misery_strategy(users_rating,num_rated_movies)
  ratings ['average without misery'] = average_without_misery_strategy(users_rating,num_rated_movies)
  top_ten_movies_index = {}
  for k,v in ratings.items():
    top_ten_movies_index [k] = v.argsort()[-10:][::-1]
  print(top_ten_movies_index)
  movies_to_rec = {}
  for k,v in top_ten_movies_index.items():
    tmp = []
    selected_movie = None
    iter = 0
    while(len(tmp)<3):
        #select the most similar movie
        selected_movie = similarities.iloc[0][rated_movies[v[iter]]]
        #if our list is not empty, check for duplications
        
        if(len(tmp)>0 and tmp.count(selected_movie)>0):
          selected_movie = None
        else:
          tmp.append(selected_movie)
        iter+=1
    
    movies_to_rec[k]=tmp
  
  groups_rec[key] = movies_to_rec  
```
